from sklearn.feature_extraction.text import TfidfVectorizer

# 示例文本
documents = [
    "text today modern societi rapid advanc technolog profoundli impact peopl lifestyl individu reli smartphon internet inform social media platform becom essenti tool commun although shift brought conveni also rais concern privaci inform secur ",
    "text contemporari world swift progress technolog significantli alter way live peopl depend smartphon internet access inform social media emerg vital mean commun chang offer conveni simultan rais issu regard privaci data secur"
]

# 创建TF-IDF向量化器
vectorizer = TfidfVectorizer()

# 进行特征提取
tfidf_matrix = vectorizer.fit_transform(documents)

# 输出特征名称和TF-IDF矩阵
print("特征名称:\n", vectorizer.get_feature_names_out())
print("TF-IDF矩阵:\n", tfidf_matrix.toarray())

from sklearn.metrics.pairwise import cosine_similarity

# 计算余弦相似度
similarity_matrix = cosine_similarity(tfidf_matrix)

# 输出相似度矩阵
print("余弦相似度矩阵:\n", similarity_matrix)
