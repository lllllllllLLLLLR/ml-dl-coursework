```python
import numpy as np

# 定义激活函数及其导数
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# 输入特征和目标值
x = np.array([[0.5,0.2,0.1]]).T  
y = np.array([[0.4]])  # 目标值（通过考试）

# 随机初始化权重和偏置 
W1 = np.array([[0.2, 0.8, 0.5], [0.1, 0.3, 0.9],[0.4, 0.7, 0.6]])  # 输入层到隐藏层权重
b1 = np.array([[0.1], [0.2], [0.3]])  # 隐藏层偏置
W2 = np.array([[0.5],[0.6], [0.7]])  # 隐藏层到输出层权重
b2 = 0.1  # 输出层偏置

# 正向传播
print("正向传播:")
z1 = np.dot(W1.T, x) + b1  # 隐藏层输入
print(f"隐藏层输入 z1: {z1.flatten()}")
a1 = sigmoid(z1)  # 隐藏层输出
print(f"隐藏层输出 a1: {a1.flatten()}")
z2 = np.dot(W2.T, a1) + b2  # 输出层输入
print(f"输出层输入 z2: {z2.flatten()}")
a2 = sigmoid(z2)  # 输出层输出
print(f"输出层输出 a2: {a2.flatten()}")

# 计算损失
loss = 0.5 * np.square(y - a2)  # 均方损失
print(f"损失: {loss.flatten()}")

# 反向传播
print("\n反向传播:")
delta2 = (a2 - y) * sigmoid_derivative(a2)  # 输出层误差
print(f"输出层误差 delta2: {delta2.flatten()}")
delta1 = np.dot(W2, delta2) * sigmoid_derivative(a1)  # 隐藏层误差
print(f"隐藏层误差 delta1: {delta1.flatten()}")

# 更新权重和偏置
learning_rate = 0.1
W2 -= learning_rate * np.dot(a1, delta2.T)  # 更新输出层权重
b2 -= learning_rate * delta2  # 更新输出层偏置
W1 -= learning_rate * np.dot(x, delta1.T)  # 更新隐藏层权重
b1 -= learning_rate * delta1  # 更新隐藏层偏置

print("\n更新后的权重和偏置:")
print("W1:", W1)
print("b1:", b1)
print("W2:", W2)
print("b2:", b2)
```

```python
import numpy as np

# 定义激活函数及其导数
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# 输入特征和目标值
x = np.array([[0.5, 0.2, 0.1]]).T  # 输入特征（3个特征）
y = np.array([[0.4]])  # 目标值

# 随机初始化权重和偏置 
W1 = np.array([[0.2, 0.8, 0.5], [0.1, 0.3, 0.9], [0.4, 0.7, 0.6]])  # 输入层到隐藏层权重
b1 = np.array([[0.1], [0.2], [0.3]])  # 隐藏层偏置
W2 = np.array([[0.5], [0.6], [0.7]])  # 隐藏层到输出层权重
b2 = 0.1  # 输出层偏置

# 正向传播
print("正向传播:")
z1 = np.dot(W1.T, x) + b1  # 隐藏层输入
print(f"隐藏层输入 z1: {z1.flatten()}")
a1 = sigmoid(z1)  # 隐藏层输出
print(f"隐藏层输出 a1: {a1.flatten()}")
z2 = np.dot(W2.T, a1) + b2  # 输出层输入
print(f"输出层输入 z2: {z2.flatten()}")
a2 = sigmoid(z2)  # 输出层输出
print(f"输出层输出 a2: {a2.flatten()}")

# 计算损失
loss = 0.5 * np.square(y - a2)  # 均方损失
print(f"损失: {loss.flatten()}")

# 反向传播
print("\n反向传播:")
delta2 = (a2 - y) * sigmoid_derivative(a2)  # 输出层误差
print(f"输出层误差 delta2: {delta2.flatten()}")
delta1 = np.dot(W2, delta2) * sigmoid_derivative(a1)  # 隐藏层误差
print(f"隐藏层误差 delta1: {delta1.flatten()}")

# 更新权重和偏置
learning_rate = 0.1  # 学习率

# 更新输出层权重和偏置
W2 -= learning_rate * np.dot(a1, delta2.T)  # 更新W2
b2 -= learning_rate * delta2.flatten()  # 更新b2

# 更新隐藏层权重和偏置
W1 -= learning_rate * np.dot(x, delta1.T)  # 更新W1
b1 -= learning_rate * delta1  # 更新b1

# 打印更新后的权重和偏置
print("\n更新后的权重和偏置:")
print("隐藏层权重 W1:")
print(W1)
print("隐藏层偏置 b1:")
print(b1)
print("输出层权重 W2:")
print(W2)
print("输出层偏置 b2:")
print(b2)
```
