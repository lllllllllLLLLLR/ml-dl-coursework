1. 基本张量操作

(1) 创建一个一维张量，对每个元素加 10

```python
import torch

# 创建一维张量
tensor_1d = torch.tensor([1, 2, 3, 4, 5])
print("Original 1D Tensor:", tensor_1d)

# 对每个元素加 10
tensor_1d_plus_10 = tensor_1d + 10
print("Tensor after adding 10:", tensor_1d_plus_10)
```

(2) 创建一个 2x3 的随机张量，计算每一列的均值

```python
# 创建 2x3 的随机张量
tensor_2x3 = torch.rand(2, 3)
print("2x3 Random Tensor:", tensor_2x3)

# 计算每一列的均值
column_means = tensor_2x3.mean(dim=0)
print("Column Means:", column_means)
```

(3) 创建两个相同大小的二维张量，进行矩阵乘法

```python
# 创建两个相同大小的 2x2 张量
tensor_a = torch.tensor([[1, 2], [3, 4]])
tensor_b = torch.tensor([[5, 6], [7, 8]])
print("Tensor A:", tensor_a)
print("Tensor B:", tensor_b)

# 进行矩阵乘法
matrix_multiplication = torch.mm(tensor_a, tensor_b)
print("Matrix Multiplication Result:", matrix_multiplication)
```

2. 计算图与反向传播

创建标量张量并计算其梯度。

```python
# 创建标量张量 x，启用梯度计算
x = torch.tensor(3.0, requires_grad=True)
print("Initial x:", x)

# 定义函数 y = 4x^3 + 2x^2 + x
y = 4 * x**3 + 2 * x**2 + x
print("Function y:", y)

# 计算 y 的梯度
y.backward()
print("Gradient of x:", x.grad)
```

4. 自动求导与梯度计算

在矩阵上执行反向传播，使用 requires_grad=True 来跟踪计算图。
假设  z = \sum(A^2) ，我们对 A 执行反向传播。A.grad 将包含每个元素的偏导数，值为 2 * A。

```python
# 创建 2x2 的张量 A，并启用 requires_grad
A = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True)
print("Tensor A:", A)

# 定义函数 z = sum(A^2)
z = (A ** 2).sum()
print("Function z:", z)

# 对 z 进行反向传播
z.backward()
print("Gradient of A:", A.grad)
```

5. 理解 requires_grad 的作用

(1) 禁用梯度计算

```python
# 创建张量 x，不启用 requires_grad
x = torch.tensor(5.0, requires_grad=False)
print("Tensor x (requires_grad=False):", x)

# 定义函数 y = 3x^2 + 4x
y = 3 * x**2 + 4 * x

# 检查 x 是否启用梯度计算
if x.requires_grad:
    y.backward()
    print("Gradient of x:", x.grad)
else:
    print("Cannot compute gradient because requires_grad is False.")
```

由于 requires_grad=False，y.backward() 将会抛出一个错误，因为 PyTorch 不会为 x 构建计算图，因此无法计算 y 的梯度。

(2) 启用梯度计算

```python
# 创建张量 x，启用 requires_grad
x = torch.tensor(5.0, requires_grad=True)
print("Tensor x (requires_grad=True):", x)

# 重新定义函数 y = 3x^2 + 4x
y = 3 * x**2 + 4 * x
print("Function y:", y)

# 计算 y 的梯度
y.backward()
print("Gradient of x:", x.grad)
```

6. 自由思考题

在模型推理阶段，为什么通常不需要启用 requires_grad？

在推理阶段，模型仅用于前向传播（预测输出），而不会执行反向传播或更新参数。因此，禁用 requires_grad 能节省显存和计算资源，使推理更高效。没有梯度跟踪的情况下，PyTorch 会在推理时省去计算图的生成和存储。
